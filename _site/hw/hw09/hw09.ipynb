### Problem 2 (24)

The groundwater under an old industrial site holds 6 units of a
chlorinated solvent (1 unit = 100 kg). The state requires that at most 1
unit remain after three years. Each year has two steps:

1.  **Treatment.** The site may pump groundwater out and treat it at the
    surface (**pump and treat**), removing up to 3 units. Removing 0, 1,
    2 or 3 units costs \$0, \$10,000, \$25,000 or \$45,000: each extra
    unit needs more wells and more pumping, so it costs more than the
    last.
2.  **Biodegradation.** Microbes then break down 1 unit of the mass left
    after treatment, with two exceptions. At 5 or 6 units the solvent is
    concentrated enough to inhibit the microbes, so nothing degrades; at
    1 unit it is too dilute to degrade further.

What remains after biodegradation is the mass at the start of next year:

| Mass left after treatment (units)          |   0 |   1 |   2 |   3 |   4 |   5 |   6 |
|:-------------------------------------------|----:|----:|----:|----:|----:|----:|----:|
| **Mass at the start of next year (units)** |   0 |   1 |   1 |   2 |   3 |   5 |   6 |

The site wants to meet the standard at the lowest total cost.

#### Problem 2.1 (4)

Trace two simple plans year by year: (a) remove 1 unit every year; (b)
treat nothing in years 1 and 2, then remove as much as allowed in year
3. Does each meet the standard, and what does each cost?
